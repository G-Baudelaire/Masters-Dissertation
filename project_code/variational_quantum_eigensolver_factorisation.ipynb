```python
import numpy as np
from qiskit import transpile
from qiskit.circuit import QuantumCircuit, ParameterVector, Parameter
from scipy.optimize import minimize
from qiskit_aer import AerSimulator  # Correct import for simulation

# Define the number of qubits
num_qubits = 5
alpha = 0.01  # CVaR threshold
shots = 1024  # Number of measurement repetitions

# Function to create Ansatz with linear CNOT entanglement
def get_ansatz():
    params = ParameterVector("θ", num_qubits)  # Parameterized angles
    qc = QuantumCircuit(num_qubits)

    # Apply RY rotations
    for i in range(num_qubits):
        qc.ry(params[i], i)

    # Apply linear entangling CNOTs
    for i in range(num_qubits - 1):
        qc.cx(i, i + 1)

    return qc, params


def get_ansatz():
    quantum_circuit = qiskit.QuantumCircuit(5)
    theta_params = []

    # First layer: H + RY
    for i in range(5):
        quantum_circuit.h(i)
        theta = Parameter(f"θ_{len(theta_params)}")
        quantum_circuit.ry(theta, i)
        theta_params.append(theta)

    # CNOTs
    for i in range(4):
        quantum_circuit.cx(i, i + 1)

    quantum_circuit.barrier()

    # Second RY layer
    for i in range(5):
        theta = Parameter(f"θ_{len(theta_params)}")
        quantum_circuit.ry(theta, i)
        theta_params.append(theta)

    # CNOTs
    for i in range(4):
        quantum_circuit.cx(i, i + 1)

    quantum_circuit.barrier()

    # Third RY layer
    for i in range(5):
        theta = Parameter(f"θ_{len(theta_params)}")
        quantum_circuit.ry(theta, i)
        theta_params.append(theta)

    # Do not measure here; add measurements only in simulation
    return quantum_circuit, theta_params

# Calculate CVaR expectation value
def calculate_CVaR(n):
    qc, params = get_ansatz()

    def cvar_expectation(theta_values):
        """Compute CVaR expectation value."""
        bound_circuit = qc.assign_parameters(dict(zip(params, theta_values)))
        bound_circuit.measure_all()

        # Simulate circuit
        backend = AerSimulator()
        transpiled_circuit = transpile(bound_circuit, backend)
        result = backend.run(transpiled_circuit, shots=shots).result()

        # Get measurement counts
        counts = result.get_counts()

        # Convert bitstrings to eigenvalues using H |p, q⟩ = (n − pq)^2 |p, q⟩
        eigenvalues = np.array([(int(k, 2) - n) ** 2 for k in counts.keys()])
        probabilities = np.array(list(counts.values())) / shots  # Normalize

        # Sort eigenvalues based on probability
        sorted_indices = np.argsort(eigenvalues)
        sorted_eigenvalues = eigenvalues[sorted_indices]
        sorted_probabilities = probabilities[sorted_indices]

        # Compute cumulative probabilities
        cumulative_probs = np.cumsum(sorted_probabilities)
        cvar_mask = cumulative_probs <= alpha  # Select lowest α fraction
        cvar_value = np.sum(sorted_eigenvalues[cvar_mask] * sorted_probabilities[cvar_mask]) / alpha

        return cvar_value

    return cvar_expectation

# Minimize CVaR using SciPy's COBYLA
def minimize_CVaR(n):
    cvar_expectation = calculate_CVaR(n)
    num_params = 5 * 3  # 3 RY layers with 5 qubits = 15 parameters
    initial_params = np.random.uniform(-np.pi, np.pi, num_params)
    result = minimize(cvar_expectation, initial_params, method="COBYLA", options={"maxiter": 200})
    return result.x

# Factorisation target
n = 15  # Example: factor 15 into 3 x 5

# Build Hamiltonian for n (if needed)
H = construct_hamiltonian(n)

# Execute the optimization
optimal_params = minimize_CVaR(n)

# Use optimal parameters to get final measurement output
final_circuit, _ = get_ansatz()
final_circuit = final_circuit.assign_parameters(optimal_params)
final_circuit.measure_all()

# Simulate final circuit
backend = AerSimulator()
transpiled_circuit = transpile(final_circuit, backend)
final_result = backend.run(transpiled_circuit, shots=shots).result()
measurement_output = final_result.get_counts()

# Check for factors (if pq = n appears in measurement output)
def bitstring_to_factors(bitstring, N_p, N_q):
    """Split bitstring into p and q, convert to int."""
    p_bits = bitstring[:N_p]
    q_bits = bitstring[N_p:N_p + N_q]
    p = int(p_bits, 2)
    q = int(q_bits, 2)
    return p, q

# Define bit widths (must match your circuit encoding)
N_p = 2
N_q = 3

# Check measurement outputs for valid factors
found_factors = False
for bitstring in measurement_output:
    p, q = bitstring_to_factors(bitstring, N_p, N_q)
    if p * q == n:
        found_factors = True
        print(f"Found valid factor pair: p = {p}, q = {q}")

if not found_factors:
    print("No valid factors found.")
# Display results
print("n =", n)
print("Optimal Parameters:", optimal_params)
print("Measurement Output:", measurement_output)
print("Maximum: ", sorted(measurement_output, key=lambda x: measurement_output[x]))
print("Found Factors:", found_factors)
```
